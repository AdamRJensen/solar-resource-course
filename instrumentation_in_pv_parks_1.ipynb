Pandas have a very convenient ["resample" function](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.resample.html) that allows for frequency conversion of time series. Note that depending on the nymber of columns, resampling the entire dataframe can be very slow. So it is more common to apply this function to the desired columns. E.g., `data[['poa_east_row','poa_west_row']].resample('1ME').sum()`.
